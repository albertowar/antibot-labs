```python
import json
key = json.load(open("../solutions/answer_key.json"))
rows = []
for n in (1, 2, 3, 4, 5):
    try:
        s = set(pd.read_csv(f"../suspects_r{n}.csv").account_id)
    except FileNotFoundError:
        continue
    for cname, c in key["campaigns"].items():
        ids = [a["account_id"] for a in c["accounts"]]
        rows.append({"round": n, "campaign": cname, "tier": c["tier"], "caught": sum(a in s for a in ids), "total": len(ids)})
pd.DataFrame(rows).pivot(index=["tier", "campaign"], columns="round", values="caught")
```
